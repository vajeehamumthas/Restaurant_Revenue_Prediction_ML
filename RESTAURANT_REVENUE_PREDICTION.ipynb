```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsRegressor
from sklearn.svm import SVR
from sklearn.metrics import mean_absolute_error,mean_squared_error,r2_score
```

```python
data=pd.read_csv('/content/drive/MyDrive/restaurant_data.csv')
data
```

```python
data.drop('Name',axis=1,inplace=True)
data.head()
```

```python
data['Location'].unique()
```

```python
data['Cuisine'].unique()
```

```python
data['Parking Availability'].unique()
```

```python
le=LabelEncoder()
data['Location']=le.fit_transform(data['Location'])
data['Cuisine']=le.fit_transform(data['Cuisine'])
data['Parking Availability']=le.fit_transform(data['Parking Availability'])
```

```python
data.dtypes
```

```python
x=data.drop('Revenue',axis=1)
y=data['Revenue']
```

```python
x_train,x_test,y_train,y_test=train=train_test_split(x,y,test_size=0.2,random_state=24)
```

#LINEAR REGRESSION


```python
lr_model=LinearRegression()
lr_model.fit(x_train,y_train)
```

```python
lr_pred=lr_model.predict(x_test)
```

```python
pd.DataFrame({'Actual':y_test,'Predicted':lr_pred})
```

```python
print('Training Score - ',lr_model.score(x_train,y_train))
print('Testing Score - ',lr_model.score(x_test,y_test))
print('Mean Absolute Error - ',mean_absolute_error(y_test,lr_pred))
print('Mean Squared Error - ',mean_squared_error(y_test,lr_pred))
print('R2 Score - ',r2_score(y_test,lr_pred))
```

#DECISION TREE REGRESSOR


```python
dt_model=DecisionTreeRegressor()
dt_model.fit(x_train,y_train)
```

```python
dt_pred=dt_model.predict(x_test)
```

```python
pd.DataFrame({'Actual':y_test,'Predicted':dt_pred})
```

```python
print('Training Score - ',dt_model.score(x_train,y_train))
print('Testing Score - ',dt_model.score(x_test,y_test))
print('Mean Absolute Error - ',mean_absolute_error(y_test,dt_pred))
print('Mean Squared Error - ',mean_squared_error(y_test,dt_pred))
print('R2 Score - ',r2_score(y_test,dt_pred))
```

#RANDOM FOREST

```python
rf_model=RandomForestRegressor()
rf_model.fit(x_train,y_train)
```

```python
rf_pred=rf_model.predict(x_test)
```

```python
pd.DataFrame({'Actual':y_test,'Predicted':rf_pred})
```

```python
print('Training Score - ',rf_model.score(x_train,y_train))
print('Testing Score - ',rf_model.score(x_test,y_test))
print('Mean Absolute Error - ',mean_absolute_error(y_test,rf_pred))
print('Mean Squared Error - ',mean_squared_error(y_test,rf_pred))
print('R2 Score - ',r2_score(y_test,rf_pred))
```
